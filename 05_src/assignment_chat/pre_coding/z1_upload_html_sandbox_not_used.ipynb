### Upload html documents from local folder

```python
import bs4      # Just checkin Beautiful Soup 4 is installed
```

```python
from langchain_community.document_loaders import BSHTMLLoader
```

BSHTMLLoader works essentially the same way as WebBaseLoader (both use BeautifulSoup for parsing), but it's designed to work with local files.  
It accepts relative paths directly without needing any URL conversion.

```python
from pathlib import Path

# Single file
relative_path = "./6k_filings/CIK0001347557_0001171843-25-007142_f6k_110725.htm"
loader = BSHTMLLoader(relative_path)        # BSHTMLLoader strips HTML tags and returns only the text content.
documents = loader.load()
```

```python
# Alternative: Read the file directly
html_content = Path(relative_path).read_text(encoding='utf-8')
# Now html_content contains the full HTML with all tags
```

```python
html_content
```

```python
# This renders the HTML directly in your Jupyter notebook cell.
from IPython.display import HTML, display
display(HTML(html_content))
```

```python
# This shows the formatted HTML source code (not rendered).
from bs4 import BeautifulSoup
soup = BeautifulSoup(html_content, 'html.parser')
print(soup.prettify())
```
